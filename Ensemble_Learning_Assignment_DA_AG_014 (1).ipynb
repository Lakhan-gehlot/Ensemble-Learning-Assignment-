# Ensemble Learning Assignment 



## Question 1
**What is Ensemble Learning?**

Ensemble Learning is a machine learning technique where multiple models are trained and combined to solve the same problem. The main idea is that combining several weak or moderately strong models leads to better accuracy, robustness, and generalization compared to a single model.

## Question 2
**Difference between Bagging and Boosting**

- Bagging reduces variance by training models independently on bootstrap samples.
- Boosting reduces bias by training models sequentially and focusing on previous errors.

## Question 3
**Bootstrap Sampling**

Bootstrap sampling randomly selects data points with replacement to create multiple training datasets. In Random Forest, it ensures diversity among trees.

## Question 4
**Out-of-Bag (OOB) Samples**

OOB samples are data points not included in a bootstrap sample. OOB score is used as an internal validation metric without needing a separate validation set.

## Question 5
**Feature Importance: Decision Tree vs Random Forest**

Decision Trees can be unstable and biased, while Random Forest averages importance across trees, making it more reliable.

```python

from sklearn.datasets import load_breast_cancer
from sklearn.ensemble import RandomForestClassifier
import pandas as pd

data = load_breast_cancer()
X, y = data.data, data.target

rf = RandomForestClassifier(random_state=42)
rf.fit(X, y)

importances = pd.Series(rf.feature_importances_, index=data.feature_names)
print(importances.sort_values(ascending=False).head(5))
```

```python

from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import BaggingClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

iris = load_iris()
X, y = iris.data, iris.target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

dt = DecisionTreeClassifier()
bag = BaggingClassifier(estimator=dt, n_estimators=50, random_state=42)

dt.fit(X_train, y_train)
bag.fit(X_train, y_train)

print("Decision Tree Accuracy:", accuracy_score(y_test, dt.predict(X_test)))
print("Bagging Accuracy:", accuracy_score(y_test, bag.predict(X_test)))
```

```python

from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier

params = {
    'n_estimators': [50, 100],
    'max_depth': [None, 5, 10]
}

grid = GridSearchCV(RandomForestClassifier(random_state=42), params, cv=5)
grid.fit(X, y)

print("Best Parameters:", grid.best_params_)
print("Best Accuracy:", grid.best_score_)
```

```python

from sklearn.datasets import fetch_california_housing
from sklearn.ensemble import BaggingRegressor, RandomForestRegressor
from sklearn.metrics import mean_squared_error

housing = fetch_california_housing()
X, y = housing.data, housing.target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

bag_reg = BaggingRegressor(random_state=42)
rf_reg = RandomForestRegressor(random_state=42)

bag_reg.fit(X_train, y_train)
rf_reg.fit(X_train, y_train)

print("Bagging MSE:", mean_squared_error(y_test, bag_reg.predict(X_test)))
print("Random Forest MSE:", mean_squared_error(y_test, rf_reg.predict(X_test)))
```
